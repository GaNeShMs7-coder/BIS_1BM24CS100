```python

import random

random.seed(7)

# Jobs: (Machine, Processing Time)
jobs = {
    1: [(0, 3), (1, 2)],
    2: [(1, 2), (0, 4)],
    3: [(0, 2), (1, 3)]
}

N, T = 10, 30
alpha, beta, rho = 1, 2, 0.5

pheromone = [[1.0]*3 for _ in range(3)]
best_cost = float('inf')

def schedule(order):
    next_op = [0]*3
    job_time = [0]*3
    machine_time = [0]*2
    result = []

    while any(x < 2 for x in next_op):
        choices = []
        for rank, j in enumerate(order):
            if next_op[j] < 2:
                m, p = jobs[j+1][next_op[j]]
                start = max(job_time[j], machine_time[m])
                choices.append((start, rank, j, m, p))

        s, _, j, m, p = min(choices)
        end = s + p
        result.append((j+1, m+1, s, end))
        job_time[j] = end
        machine_time[m] = end
        next_op[j] += 1

    return max(job_time), result

# ACO optimization
for t in range(T):
    for _ in range(N):
        order = [random.randrange(3)]

        while len(order) < 3:
            current = order[-1]
            remaining = [j for j in range(3) if j not in order]
            weights = [pheromone[current][j]**alpha /
                       (sum(p for m, p in jobs[j+1])**beta)
                       for j in remaining]
            order.append(random.choices(remaining, weights)[0])

        cost, result = schedule(order)

        if cost < best_cost:
            best_cost, best_order, best_result = cost, order[:], result

    for i in range(3):
        for j in range(3):
            pheromone[i][j] *= (1-rho)

    for i in range(len(order)-1):
        pheromone[order[i]][order[i+1]] += 1/cost

# Display output
print("===== ACO JOB SHOP SCHEDULING =====")
print("Optimal Job Sequence:", [j+1 for j in best_order])
print("\nJob  Machine  Start  Finish")

for j, m, s, e in best_result:
    print(f"J{j}      M{m}       {s}      {e}")

print("\nMinimum Makespan:", best_cost)
print("Maximum Iterations:", T)
print("Optimization: SUCCESS")
print("===================================")
```
