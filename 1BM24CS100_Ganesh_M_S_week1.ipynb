```python

demand = int(input("Enter demand: "))
ordering_cost = float(input("Enter ordering cost: "))
holding_cost = float(input("Enter holding cost per unit: "))
shortage_cost = float(input("Enter shortage cost per unit: "))

min_cost = float('inf')
best_quantity = 0

for quantity in range(1, demand * 2 + 1):

    order_cost = ordering_cost

    hold_cost = max(0, quantity - demand) * holding_cost

    short_cost = max(0, demand - quantity) * shortage_cost

    total_cost = order_cost + hold_cost + short_cost

    print("Quantity:", quantity, "Total Cost:", total_cost)

    if total_cost < min_cost:
        min_cost = total_cost
        best_quantity = quantity

print("\nBest Inventory Quantity:", best_quantity)
print("Minimum Total Cost:", min_cost)
```
